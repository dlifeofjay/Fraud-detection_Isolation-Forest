# Goal: Find out the factors the makes transaction fraud

```python
# Data Standardization and Cleaning

df["TransactionDate"] = pd.to_datetime(df["TransactionDate"], format="mixed")

df = df.sort_values(by="TransactionDate")
```

## Goal Assumption: 
- Fraudulent transactions are determined by activities done that led to that transaction being made.
- We dont need customer profiles or demographics, we will need them to determine whe profile of customers after we have determined what makes transaction fraudulent.
- Fraudulent transactions are expected not to follow general transaction behaviour
- Both credit and debit transactions can be fraudulent e.g Sudden large debit or money laundering

## Sub-Goal: Lets determine what makes transaction fraudulent and create a baseline

```python
df.columns
```

```python
df_tot = df[["AccountID", "TransactionDate", "TransactionID", "TransactionAmount", "TransactionType", "TransactionDuration", "LoginAttempts",
    "AccountBalance"]]
```

# Start with debit transactions

```python
df1 = df_tot[df_tot["TransactionType"] == "Debit"]
```

```python
# Create amount debited to balance rate assuming balance is predebit

df1["Amount/Balance"] = df1.apply(lambda x: x["TransactionAmount"] / x["AccountBalance"], axis=1)
```

```python
df1.dropna(inplace=True)
```

```python
# View correlation
df_num = df1.select_dtypes(include=["int", "float"])

plt.figure(figsize=(7, 5))
sns.heatmap(df_num.corr(), annot=True)
plt.title("Correlation Matrix")
plt.show()
```

**No strong relationship between variables above**

```python
# View distribution to view skewness

fig, axes = plt.subplots(3, 2, figsize=(17, 15))

# For loop for easy plots
for ax, col in zip(axes.flatten(), df_num.columns):
    sns.histplot(df_num[col], kde=True, ax=ax)
    ax.set_title(col)

plt.tight_layout()
plt.show()
```

## Note
- Transaction Amount and login are right skewed, meaning most transactions consist of lower values and most login attempts are 1
- Transaction duration and account balance are also mostly lower values but higher values still have momentum.
- Amount per balance rate is right skewed indicating most individuals has high account balance in relation to the transaction done by them

```python
# Lets find the outlier values in this variables using IQR method
Q1 = df_num.quantile(0.25)
Q3 = df_num.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_mask = ((df_num < lower_bound) | (df_num > upper_bound)).any(axis=1)

df_out = df_num[outlier_mask]

df_nor = df_num[~outlier_mask]
```

```python
# Average values of outlier transaction
df_out.median()
```

```python
# Average values of normal transactions
df_nor.median()
```

```python
max_nor = df_nor["Amount/Balance"].max()
max_out = df_out["Amount/Balance"].max()

print(f"Maximum Rate of Amount to Balance for Normal Transaction: {max_nor}")
print(f"Maximum Rate of Amount to Balance for Outlier Transaction: {max_out}")
```

# Above shows
- Main difference between these 2 types of transactions are transaction amount and account balance, outlier transaction amount is much higher than that of normal transaction while the account balance is much lower
- This makes Amount/Balance rate for both parties different
- There are cases where successful amount debited is more than current account balance in outlier transactions, sign of fraud

**Viewing distribution for clarity**

```python
# for outlier first
fig, axes = plt.subplots(3, 2, figsize=(17, 15))

fig.suptitle("Outlier Transaction Distribution", fontsize=20)

for ax, col in zip(axes.flatten(), df_out.columns):
    sns.histplot(df_out[col], kde=True, ax=ax)
    ax.set_title(col)

plt.tight_layout()
plt.show()
```

```python
# for normal first
fig, axes = plt.subplots(3, 2, figsize=(17, 15))

fig.suptitle("Normal Transaction Distribution", fontsize=20)

for ax, col in zip(axes.flatten(), df_nor.columns):
    sns.histplot(df_nor[col], kde=True, ax=ax)
    ax.set_title(col)

plt.tight_layout()
plt.show()
```

## Note
- Outlier transactions are having transactions up to 2,000 while normal is having roughly 1,000 as highest
- Login counts for outlier transactions are up to 5 while for the normal transactions has just 1 login attempt
- The key observation is the amount made per balance as outlier transaction is getting up to 80% of account balance being withdrawn
- No real difference in transaction duration and account balance

**Lets view the relationship**

```python
# Correlation for outlier transactions

plt.figure(figsize=(7, 5))
sns.heatmap(df_out.corr(), annot=True)
plt.title("Correlation Matrix For Outlier")
plt.show()
```

**I need to find case where we have high login rate at least more 3 or more than and high rate of amount/balance if its one time large debit fraud**

```python
# Distribution of Login Attempts and Amount per balance for Outlier Transactions

plt.figure(figsize=(7, 5))
sns.boxplot(data=df_out, x="LoginAttempts", y="Amount/Balance")
plt.title("Relationship Between Login Attempts and Transaction Amount")
plt.show()
```

- Above shows outlier transactions with 1 login attempts carries different different amount rate per balance, logins with 5 has less than 20% rate

**Normal transactions have just 1 login counts and highest amount per balance rate is 35%**

**Since we high login attempt doesnt mean high amount  per balance rate look for a scenerio where first login has high login attempt but repeat withdrawal in that same day**

```python
# Lets look for account id that login attempt is 3 and above at least once and also did multiple transaction in same day
cols = df_out.columns

result = ["AccountID", "TransactionDate"] + cols.tolist()
new_df = df1[result]

# Filter
suspicious_accounts = []

for account, group in new_df.groupby('AccountID'):
    has_high_login = (group['LoginAttempts'] >= 3).any()
    has_multiple_txns_same_day = group['TransactionDate'].dt.date.duplicated().any()
    
    if has_high_login and has_multiple_txns_same_day:
        suspicious_accounts.append(account)
```

```python
new_df[(new_df["AccountID"].isin(suspicious_accounts))]
```
